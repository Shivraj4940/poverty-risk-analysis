```python
#Remove First Column (Timestamp) 
df=df.iloc[:,1:]
```

```python
# Change Columns name for analysis 
name=['Age_group', 'Gender', 'Household_size', 'Family_type', 'Residence_area', 'Occupation_of_main_earner', 'Highest_education_in_household', 'Ration_Card_type', 'Type_of_hospital', 'Type_of_patient', 'Type_of_illness', 'Visits_per_month', 'Hospital_visit_cost', 'Consultation_fee', 'Medicine_cost', 'Diagnostics_cost', 'Travel_cost', 'Lost_income', '3_month_total_expenditure', 'Monthly_income', 'Borrowed_money', 'Financial_hardship', 'Financial_stability', 'Insurance_enrolled', 'Insurance_used', 'Financial_support_level', 'Ease_of_access', 'Trust_level']
df.columns = name
```

```python
df.head()
```

```python
# Function For replace unwanted symbol 
def replace_dash(x):
    try:
        x = str(x)
        return x.replace("–", "/").replace("â€“", "/")
    except:
        return x

# Apply Fucntion 
df['Occupation_of_main_earner'] = df['Occupation_of_main_earner'].apply(replace_dash)
```

```python
# Replace value from specific row and column together 
df.loc[
    df['Type_of_hospital'] == 'Government hospital',
    ['Hospital_visit_cost', 'Consultation_fee', 'Diagnostics_cost']
] = 0
```

```python
#Cleaning Special Characters and Bracket Text
import re

def clean(x):
    try:
        x = str(x)
        x = x.split("/")[0]              
        x = re.sub(r"\s*\(.*?\)", "", x)    
        return x.strip()
    except:
        return x
```

```python
# Apply function 
for col in df.columns:
    df[col] = df[col].apply(clean)
```

```python
# Function for Removing unwanted numbers/symbols
import re

def clean_illness(x):
    try:
        x = str(x)
        
        # Fix encoding (important)
        try:
            x = x.encode('latin1').decode('utf-8')
        except:
            pass
        
        # Keep only English words at beginning
        match = re.match(r'^[A-Za-z\s\-]+', x)
        if match:
            return match.group().strip()
        
        return x.strip()
    
    except:
        return x

df['Type_of_illness'] = df['Type_of_illness'].apply(clean_illness)
```

```python
df.head()
```

```python
df = df.drop(df[df['Gender'] == 'Other'].index)
```

```python
#Converting Text Ranges into Numeric Values 
import re

def convert_value(x):
    try:
        x = str(x)
        
        # Remove commas first
        x = x.replace(",", "")
        
        # Case 1: More than 5000 → 5000
        if "More than" in x:
            return int(re.findall(r"\d+", x)[0])
        
        # Case 2: 0-500 → 500
        if "-" in x:
            return int(x.split("-")[1])
        
        # Case 3: already number
        return int(x)
    
    except:
        return x

df['3_month_total_expenditure']=df['3_month_total_expenditure'].apply(convert_value)
```

```python
#change data type
df['3_month_total_expenditure'] = pd.to_numeric(
    df['3_month_total_expenditure'], errors='coerce'
)
```

```python
#check for missing value 
df.isnull().sum()
```

```python
#fill missing value
df['3_month_total_expenditure']=df['3_month_total_expenditure'].fillna(df['3_month_total_expenditure'].mean())
df['Ease_of_access']=df['Ease_of_access'].fillna(df['Ease_of_access'].mode()[0])
```

```python
# convert columns into numeric 
import re

def convert_to_number(x):
    try:
        x = str(x).strip().lower()
        
        # Yes / No
        if x in ["yes", "1. yes", "yes.", "y"]:
            return 1
        if x in ["no", "2. no", "no.", "n","nan","Not applicable"]:
            return 0
        
        # Fix encoding issues
        x = x.replace("â€“", "-").replace("–", "-").replace("--", "-")
        
        # Case: More than / Less than (handle comma numbers correctly)
        if "more than" in x or "less than" in x:
            num = re.findall(r"[\d,]+", x)   # capture full number like 11,500
            if num:
                return float(num[0].replace(",", ""))
        
        # Remove unwanted characters except digits and -
        x = re.sub(r"[^\d\-,]", "", x)
        
        # Remove commas
        x = x.replace(",", "")
        
        # Case: Range → midpoint
        if "-" in x:
            parts = [p for p in x.split("-") if p != ""]
            if len(parts) == 2:
                return (float(parts[0]) + float(parts[1])) / 2
        
        # Case: Single number
        if x.isdigit():
            return float(x)
        
        return 0
    
    except:
        return 0
```

```python
#New data frame for numeric column 
cols=['Household_size','Visits_per_month', 'Hospital_visit_cost', 'Consultation_fee', 'Medicine_cost', 'Diagnostics_cost', 'Travel_cost','Monthly_income','Lost_income','Borrowed_money', 'Financial_hardship', 'Financial_stability', 'Insurance_enrolled', 'Insurance_used', 'Financial_support_level', 'Ease_of_access', 'Trust_level']
new_df = pd.DataFrame()

for col in cols:
    new_df[col + "_numeric"] = df[col].apply(convert_to_number)
```

```python
#combination of data frame 
cols = [
    'Age_group', 'Gender', 'Household_size', 'Family_type',
    'Residence_area', 'Occupation_of_main_earner',
    'Highest_education_in_household', 'Ration_Card_type',
    'Type_of_hospital', 'Type_of_patient', 'Type_of_illness','3_month_total_expenditure'
]

df_selected = df[cols]

final_df = pd.concat([df_selected, new_df], axis=1)
print(final_df.head())
print(final_df.shape)
```

```python
#Convert float value into integer
final_df['Visits_per_month_numeric']=final_df['Visits_per_month_numeric'].round().astype(int)
```

```python
# Fill 0 value with mode value
mode_val = final_df['3_month_total_expenditure'][final_df['3_month_total_expenditure'] != 0].mode()[0]

final_df['3_month_total_expenditure'] = final_df['3_month_total_expenditure'].replace(0, mode_val)
```

```python
# Save dataframe into csv file 
final_df.to_csv("Hospital Data.csv", index=False)
```
